# Spam Detector



**This Cell will  ''' Import '''  the Required Librarires**

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report
```

**Load Dataset (replace with your dataset file path)**


**Dataset should have two columns: 'text' and 'label' (spam or ham)**

```python
data = pd.read_csv('/content/sample_data/spam_dataset.csv')
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

**Splitting the dataset**

```python
X = data['Message']  # Email content
y = data['Category']  # Spam or ham
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.35, random_state=42)
```

**Build Pipeline**

```python
pipeline = Pipeline([
    ('vectorizer', TfidfVectorizer(stop_words='english')),  # Text vectorization
    ('classifier', MultinomialNB())  # Naive Bayes classifier
])
```

**Train the model**

```python
pipeline.fit(X_train, y_train)
```

**Evaluate the model**

```python
y_pred = pipeline.predict(X_test)
print("Model Evaluation:")
print(classification_report(y_test, y_pred))
```

**Interactive Input**

```python
print("\nModel is ready! Enter a message to classify it as spam or not spam.")

while True:
    user_input = input("\nEnter text (or type 'exit' to quit): ")
    if user_input.lower() == 'exit':
        print("Exiting the program.")
        break
    # prediction = pipeline.predict([user_input])[0]
    # print(f"The message is classified as: {'SPAM' if prediction == 'spam' else 'NOT SPAM'}")
    prediction = pipeline.predict([user_input])[0]
    prediction_proba = pipeline.predict_proba([user_input])
    print(f"The message is classified as: {'SPAM' if prediction == 'spam' else 'NOT SPAM'}")
    print(f"Spam probability: {prediction_proba[0][1]:.2f}, Ham probability: {prediction_proba[0][0]:.2f}")
```
